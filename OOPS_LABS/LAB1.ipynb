```python
numbers = [10, 20, 30, 40, 50]

numbers = numbers[-1:] + numbers[:-1]

print(numbers)
```

```python
numbers = [10, 20, 30, 40, 50]

numbers[2] = numbers[0] + numbers[-1]

print(numbers)
```

```python
A = {1, 2, 3}
B = {4, 5, 6}

if A.isdisjoint(B):
    print("Union:", A | B)
else:
    print("Sets have common elements")
```

```python
A = {1, 2}
B = {1, 2, 3, 4}

if A.issubset(B):
    print("Subset")
else:
    print("Not Subset")
```

```python
A = {1, 2, 3}
B = {3, 4, 5}

common = A & B

if common:
    print("Common Found")
else:
    print("No Common")
```

```python
numbers = (10, 20, 30, 40, 50)

numbers = (numbers[-1],) + numbers[1:-1] + (numbers[0],)

print(numbers)
```

```python
class_A = (1, 2, 3, 4)
class_B = (3, 4, 5, 6)

common = set(class_A) & set(class_B)

print(common)
```

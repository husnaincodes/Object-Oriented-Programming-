Question no 3

```python
class BankAccount:
    def __init__(self, account_holder, balance):
        self.account_holder = account_holder
        self.balance = balance

    def deposit(self, amount):
        self.balance = self.balance + amount


class SavingsAccount(BankAccount):
    def add_interest(self):
        self.balance = self.balance + (self.balance * 0.05)


account1 = SavingsAccount("Husnain", 60000)

account1.deposit(10000)
account1.add_interest()

print("Account Holder:", account1.account_holder)
print("Final Balance:", account1.balance)
```

Question no 6

```python
class Employee:
    def __init__(self, name, basic_salary):
        self.name = name
        self.basic_salary = basic_salary


class SalesEmployee(Employee):
    def __init__(self, name, basic_salary, sales):
        self.name = name
        self.basic_salary = basic_salary
        self.sales = sales

    def final_salary(self):
        if self.sales >= 50000:
            bonus = self.basic_salary * 0.10
        else:
            bonus = 0

        return self.basic_salary + bonus


employee1 = SalesEmployee("Rafay", 80000, 60000)

print("Name:", employee1.name)
print("Basic Salary:", employee1.basic_salary)
print("Sales:", employee1.sales)
print("Final Salary:", employee1.final_salary())
```

Question no 7

```python
class Order:
    def __init__(self, customer_name, total_amount):
        self.customer_name = customer_name
        self.total_amount = total_amount

    def display_order(self):
        print("Customer Name:", self.customer_name)
        print("Total Amount:", self.total_amount)


class OnlineOrder(Order):
    def __init__(self, customer_name, total_amount, delivery_fee):
        self.customer_name = customer_name
        self.total_amount = total_amount
        self.delivery_fee = delivery_fee

    def final_bill(self):
        return self.total_amount + self.delivery_fee

    def __del__(self):
        print("Online order deleted")


order1 = OnlineOrder("Shahbaz", 5000, 300)

order1.display_order()
print("Delivery Fee:", order1.delivery_fee)
print("Final Bill:", order1.final_bill())

del order1
```

Question no 9

```python
class Vehicle:
    def __init__(self, brand):
        self.brand = brand


class SpeedChecker(Vehicle):
    def __init__(self, brand, speed):
        self.brand = brand
        self.speed = speed

    def check_speed(self):
        if self.speed > 120:
            return "Overspeeding"
        elif self.speed > 60:
            return "Normal Speed"
        else:
            return "Slow Speed"


vehicle1 = SpeedChecker("BMW", 100)

print("Brand:", vehicle1.brand)
print("Speed:", vehicle1.speed)
print("Speed Category:", vehicle1.check_speed())
```

Question no 10

```python
class Product:
    def __init__(self, name):
        self.name = name


class Inventory(Product):
    def __init__(self, name, stock):
        self.name = name
        self.stock = stock

    def display_stock(self):
        total = 0

        for quantity in self.stock:
            print("Stock Quantity:", quantity)
            total = total + quantity

        print("Product Name:", self.name)
        print("Total Stock:", total)


inventory1 = Inventory("Laptop", [10, 20, 15, 5])

inventory1.display_stock()
```

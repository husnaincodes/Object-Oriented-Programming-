```python
class Employee:
    def __init__(self, name, emp_id):
        self.name = name
        self.emp_id = emp_id

class DepartmentEmployee(Employee):
    def __init__(self, name, emp_id, department):
        super().__init__(name, emp_id)
        self.department = department

    def display_info(self):
        print("Name:", self.name)
        print("ID:", self.emp_id)
        print("Department:", self.department)

e = DepartmentEmployee("Husnain", "E101", "Finance")
e.display_info()
```

```python
class Customer:
    def __init__(self, name, phone):
        self.name = name
        self.phone = phone

class PremiumCustomer(Customer):
    def __init__(self, name, phone, years):
        super().__init__(name, phone)
        self.years = years

    def display_info(self):
        print("Name:", self.name)
        print("Phone:", self.phone)
        print("Years:", self.years)
        if self.years >= 3:
            print("Loyal Customer")
        else:
            print("Regular Customer")

c = PremiumCustomer("Dua", "031477141714", 4)
c.display_info()
```

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

class StockedProduct(Product):
    def __init__(self, name, price, stock):
        super().__init__(name, price)
        self.stock = stock

    def display_stock(self):
        print("Stock quantities:", self.stock)
        total = sum(self.stock)
        print("Total Stock:", total)
        if total > 100:
            print("Sufficient Stock")
        else:
            print("Low Stock")

p = StockedProduct("PC", 80000, [60, 35, 50])
p.display_stock()
```

```python
class Patient:
    def __init__(self, name, patient_id):
        self.name = name
        self.patient_id = patient_id

class MedicalRecord(Patient):
    def __init__(self, name, patient_id, temps):
        super().__init__(name, patient_id)
        self.temps = temps

    def check(self):
        print("Temperatures:", self.temps)
        avg = sum(self.temps) / len(self.temps)
        print("Average:", avg)
        if avg > 37.5:
            print("Require Attention")
        else:
            print("Normal")

r = MedicalRecord("Rafay", "Patient201", [36.8, 37.9, 38.1, 37.2])
r.check()
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

class UniversityStudent(Person):
    def __init__(self, name, age, sid, program):
        super().__init__(name, age)
        self.sid = sid
        self.program = program

    def display(self):
        print("Name:", self.name)
        print("Age:", self.age)
        print("ID:", self.sid)
        print("Program:", self.program)

s1 = UniversityStudent("Shahbaz", 20, "BSFT-10", "Software Engineering")
s2 = UniversityStudent("Aliya", 21, "BSFT-20", "Data science")
s1.display()
s2.display()
```

```python
class Vehicle:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

class ElectricVehicle(Vehicle):
    def __init__(self, brand, model, capacity, percent):
        super().__init__(brand, model)
        self.capacity = capacity
        self.percent = percent

    def check(self):
        energy = self.capacity * self.percent / 100
        print("Brand:", self.brand)
        print("Model:", self.model)
        print("Available Energy:", energy)
        if energy > 30:
            print("Sufficient Battery")
        else:
            print("Low Battery")

ev = ElectricVehicle("TOYOTA", "Model 5", 50, 80)
ev.check()
```

```python
class Course:
    def __init__(self, name, fee):
        self.name = name
        self.fee = fee

class OnlineCourse(Course):
    def __init__(self, name, fee, lectures):
        super().__init__(name, fee)
        self.lectures = lectures

    def display(self):
        cost = self.fee / self.lectures
        print("Course:", self.name)
        print("Fee:", self.fee)
        print("Lectures:", self.lectures)
        print("Cost per Lecture:", cost)

c = OnlineCourse("Python OOP", 18000, 20)
c.display()
```

```python
class Sensor:
    def __init__(self, sid, location):
        self.sid = sid
        self.location = location

class TemperatureSensor(Sensor):
    def __init__(self, sid, location, temps):
        super().__init__(sid, location)
        self.temps = temps

    def analyze(self):
        print("Temperatures:", self.temps)
        print("Sensor ID:", self.sid)
        print("Location:", self.location)
        print("Highest:", max(self.temps))
        print("Lowest:", min(self.temps))

s = TemperatureSensor("S061", "Lab", [22.5, 25.0, 21.8, 27.3, 23.1])
s.analyze()
```

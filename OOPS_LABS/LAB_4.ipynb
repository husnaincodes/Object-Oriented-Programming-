```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def display_student(self):
        print(self.name, self.marks)


class Course:
    def __init__(self, course_name):
        self.course_name = course_name
        self.students = []

    def add_student(self, student):
        self.students.append(student)

    def display_students(self):
        print("Course:", self.course_name)

        for student in self.students:
            student.display_student()

    def calculate_average(self):
        total = 0

        for student in self.students:
            total = total + student.marks

        average = total / len(self.students)
        print("Average:", average)


s1 = Student("Ramdam", 80)
s2 = Student("Shaheer", 90)
s3 = Student("Haris", 70)

course = Course("Computer Science")

course.add_student(s1)
course.add_student(s2)
course.add_student(s3)

course.display_students()
course.calculate_average()
```

```python
class OnlineOrder:
    def __init__(self, customer_name, product_name, quantity, price):
        self.customer_name = customer_name
        self.product_name = product_name
        self.quantity = quantity
        self.price = price
        print(f"Order created for {self.customer_name}")

    def calculate_total(self):
        return self.quantity * self.price

    def display_order(self):
        print(f"Customer Name: {self.customer_name}")
        print(f"Product Name: {self.product_name}")
        print(f"Quantity: {self.quantity}")
        print(f"Price: {self.price}")
        print(f"Total Price: {self.calculate_total()}\n")

    def __del__(self):
        print(f"Order deleted for {self.customer_name}")



order1 = OnlineOrder("Ahmad", "Laptop", 1, 1200)
order2 = OnlineOrder("Rafay", "Headphones", 2, 150)


print("\nDisplaying Orders")
order1.display_order()
order2.display_order()


print("Deleting First Order")
del order1
```

```python
class Student:

    def __init__(self, name, mark1, mark2, mark3):
        self.name = name
        self.mark1 = mark1
        self.mark2 = mark2
        self.mark3 = mark3

    def average(self):
        return (self.mark1 + self.mark2 + self.mark3) / 3


class Course:

    def __init__(self, course_name):
        self.course_name = course_name
        self.students = []

    def add_student(self, student):
        self.students.append(student)

    def display_students(self):
        for student in self.students:
            print("Name:", student.name)
            print("Marks:", student.mark1, student.mark2, student.mark3)
            print("Average:", student.average())
            print()


s1 = Student("Ali", 80, 70, 90)
s2 = Student("Ahmed", 75, 85, 80)
s3 = Student("Sara", 90, 95, 85)

course = Course("Python")

course.add_student(s1)
course.add_student(s2)
course.add_student(s3)

course.display_students()
```

```python
class Product:

    def __init__(self, name, price, quantity):
        self.name = name
        self.price = price
        self.quantity = quantity


class ShoppingCart:

    def __init__(self, customer_name):
        self.customer_name = customer_name
        self.products = []

    def add_product(self, product):
        self.products.append(product)

    def calculate_bill(self):
        total = 0

        for product in self.products:
            total = total + product.price * product.quantity

        return total


p1 = Product("Pen", 50, 2)
p2 = Product("Notebook", 200, 3)
p3 = Product("Bag", 1500, 1)

cart = ShoppingCart("Mohib")

cart.add_product(p1)
cart.add_product(p2)
cart.add_product(p3)

print("Customer:", cart.customer_name)
print("Total Bill:", cart.calculate_bill())
```

```python
class Customer:

    def __init__(self, name):
        self.name = name


class BankAccount:

    def __init__(self, account_number, balance):
        self.account_number = account_number
        self.balance = balance

    def deposit(self, amount):
        self.balance = self.balance + amount

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance = self.balance - amount
            print("Withdrawal successful")
        else:
            print("Insufficient Balance")

    def display_balance(self):
        print("Balance:", self.balance)


customer = Customer("Muhmamad")

account = BankAccount("12345", 10000)

account.deposit(5000)

account.withdraw(3000)

account.withdraw(15000)

account.display_balance()
```

```python
class FoodItem:

    def __init__(self, name, price, quantity):
        self.name = name
        self.price = price
        self.quantity = quantity


class Order:

    def __init__(self, customer_name):
        self.customer_name = customer_name
        self.items = []

    def add_item(self, item):
        self.items.append(item)

    def calculate_bill(self):
        subtotal = 0

        for item in self.items:
            subtotal = subtotal + item.price * item.quantity

        if subtotal >= 5000:
            discount = subtotal * 0.10
        else:
            discount = 0

        final_bill = subtotal - discount

        print("Customer:", self.customer_name)
        print("Subtotal:", subtotal)
        print("Discount:", discount)
        print("Final Bill:", final_bill)


f1 = FoodItem("Pizza", 2000, 2)
f2 = FoodItem("Burger", 800, 2)
f3 = FoodItem("Fries", 500, 1)

order = Order("Muhammad")

order.add_item(f1)
order.add_item(f2)
order.add_item(f3)

order.calculate_bill()
```

```python
class Customer:

    def __init__(self, name):
        self.name = name


class Product:

    def __init__(self, name, price, quantity):
        self.name = name
        self.price = price
        self.quantity = quantity

    def total_price(self):
        return self.price * self.quantity


class Order:

    def __init__(self, customer):
        self.customer = customer
        self.products = []

    def add_product(self, product):
        self.products.append(product)

    def display_order(self):

        total = 0

        print("Customer:", self.customer.name)

        for product in self.products:
            print("Product:", product.name)
            print("Price:", product.price)
            print("Quantity:", product.quantity)

            total = total + product.total_price()

        print("Final Bill:", total)

        if total >= 10000:
            print("Eligible for Discount")
        else:
            print("No Discount")

    def __del__(self):
        print("Order deleted for", self.customer.name)


customer = Customer("Ahmad")

p1 = Product("Laptop", 50000, 1)
p2 = Product("Mouse", 2000, 2)
p3 = Product("Keyboard", 3000, 1)

order = Order(customer)

order.add_product(p1)
order.add_product(p2)
order.add_product(p3)

order.display_order()

del order
```

```python
class Student:

    def __init__(self, name, student_id):
        self.name = name
        self.student_id = student_id


class Course:

    def __init__(self, course_name, capacity):
        self.course_name = course_name
        self.capacity = capacity
        self.students = []

    def register_student(self, student):

        if len(self.students) < self.capacity:
            self.students.append(student)
            print(student.name, "registered successfully")
        else:
            print("Course Full")

    def display_students(self):

        print("Registered Students:")

        for student in self.students:
            print(student.name, "-", student.student_id)


s1 = Student("Ali", "001")
s2 = Student("Ahmed", "002")
s3 = Student("Sara", "003")
s4 = Student("Hamza", "004")
s5 = Student("Ayesha", "005")

course = Course("OOP", 3)

course.register_student(s1)
course.register_student(s2)
course.register_student(s3)
course.register_student(s4)
course.register_student(s5)

course.display_students()
```

```python
class Employee:

    def __init__(self, name, basic_salary, performance_rating):
        self.name = name
        self.basic_salary = basic_salary
        self.performance_rating = performance_rating


class Company:

    def __init__(self, company_name):
        self.company_name = company_name
        self.employees = []

    def add_employee(self, employee):
        self.employees.append(employee)

    def display_salary(self):

        for employee in self.employees:

            if employee.performance_rating >= 4:
                bonus = employee.basic_salary * 0.20

            elif employee.performance_rating >= 3:
                bonus = employee.basic_salary * 0.10

            else:
                bonus = 0

            final_salary = employee.basic_salary + bonus

            print("Name:", employee.name)
            print("Basic Salary:", employee.basic_salary)
            print("Bonus:", bonus)
            print("Final Salary:", final_salary)
            print()


e1 = Employee("Ali", 50000, 4)
e2 = Employee("Ahmed", 60000, 3)
e3 = Employee("Sara", 70000, 2)

company = Company("ABC Company")

company.add_employee(e1)
company.add_employee(e2)
company.add_employee(e3)

company.display_salary()
```

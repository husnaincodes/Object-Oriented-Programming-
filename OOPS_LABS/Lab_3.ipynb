**Question NO 1**

```python
class OnlineStore:
    def __init__(self, customer_name, product_name, quantity, price_per_item, available_stock):
        self.customer_name = customer_name
        self.product_name = product_name
        self.quantity = quantity
        self.price_per_item = price_per_item
        self.available_stock = available_stock

    def display_order(self):
        print("Customer Name:", self.customer_name)
        print("Product Name:", self.product_name)
        print("Quantity Purchased:", self.quantity)
        print("Price Per Item: Rs.", self.price_per_item)
        print("Total Order Amount: Rs.", self.calculate_total())

    def calculate_total(self):
        return self.quantity * self.price_per_item

    def check_stock(self):
        return self.available_stock >= self.quantity

order = OnlineStore("Ali", "Laptop", 3, 25000, 5)

order.display_order()

print("Total Amount:", order.calculate_total())

if order.check_stock():
    print("Order Available")
else:
    print("Out of Stock")

if order.calculate_total() >= 50000:
    print("Discount Applied")
else:
    print("No Discount")
```

**Question no 2**

```python
class LibraryBook:
    def __init__(self, title, author):
        self.title = title
        self.author = author

    def display_book(self):
        print("Book Title:", self.title)
        print("Author:", self.author)

    def __del__(self):
        print("Book record deleted for", self.title)

book = LibraryBook("Python Programming", "John Smith")
book.display_book()

del book
```

**Question no 3**

```python
class ShoppingOrder:
    def __init__(self, customer_name, product_name, quantity, price_per_item):
        self.customer_name = customer_name
        self.product_name = product_name
        self.quantity = quantity
        self.price_per_item = price_per_item
        self.total_price = 0

    def display_order(self):
        print("Customer Name:", self.customer_name)
        print("Product Name:", self.product_name)
        print("Quantity:", self.quantity)
        print("Total Price: Rs.", self.total_price)

    def calculate_price(self):
        self.total_price = self.quantity * self.price_per_item
        return self.total_price

    def __del__(self):
        print("Order for", self.customer_name, "has been removed.")

order = ShoppingOrder("Ali", "Wireless Headphones", 2, 5000)
order.calculate_price()
order.display_order()

del order
```

**Question no 7**

```python
class BankAccount:
    def __init__(self, name, balance):
        self.name = name
        self.__balance = balance

    def deposit(self, amount):
        self.__balance += amount

    def __check_balance(self):
        return self.__balance >= 10000

    def display_status(self):
        if self.__check_balance():
            print("Sufficient Balance")
        else:
            print("Low Balance")

    def display_info(self):
        print("Account Holder:", self.name)
        print("Current Balance: Rs.", self.__balance)

acc = BankAccount("Ali", 8000)

acc.deposit(5000)

acc.display_info()

acc.display_status()
```

**Question no 8**

```python
class Student:
    def __init__(self, name, student_id):
        self.name = name
        self.student_id = student_id

    def display_details(self):
        print("Student Name:", self.name, "| ID:", self.student_id)

class Course:
    def __init__(self, course_name):
        self.course_name = course_name
        self.students = []

    def add_student(self, student):
        self.students.append(student)

    def display_course(self):
        print("Course Name:", self.course_name)
        print("Enrolled Students:")
        for student in self.students:
            student.display_details()

s1 = Student("Ali", "FT-001")
s2 = Student("Sara", "FT-002")
s3 = Student("Ahmed", "FT-003")

course = Course("Object Oriented Programming")

course.add_student(s1)
course.add_student(s2)
course.add_student(s3)

course.display_course()
```

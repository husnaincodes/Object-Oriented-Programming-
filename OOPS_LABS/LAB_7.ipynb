```python

class Student:
    def __init__(self, name, roll_no):
        self.name = name
        self.roll_no = roll_no


class Undergraduate(Student):
    def __init__(self, name, roll_no, marks):
        super().__init__(name, roll_no)
        self.marks = marks

    def display(self):
        print("Name:", self.name)
        print("Roll Number:", self.roll_no)
        print("Marks:", self.marks)

        if self.marks >= 50:
            print("Pass")
        else:
            print("Fail")

        print()


class Graduate(Student):
    def __init__(self, name, roll_no, marks):
        super().__init__(name, roll_no)
        self.marks = marks

    def display(self):
        print("Name:", self.name)
        print("Roll Number:", self.roll_no)
        print("Marks:", self.marks)

        if self.marks >= 50:
            print("Pass")
        else:
            print("Fail")

        print()


u1 = Undergraduate("Ali", 101, 75)
u2 = Undergraduate("Ahmed", 102, 40)
g1 = Graduate("Sara", 201, 85)

u1.display()
u2.display()
g1.display()
```

```python
class A:
  def __init__(self):
    self.x = 2
    print("A:", self.x)
  def show(self):
    self.x += 3
    print("A show:", self.x)
class B(A):
  def __init__(self):
    super().__init__()
    self.x *= 2
    print("B:", self.x)
def show(self):
  print("B show:", self.x)
  super().show()
class C(A):
  def __init__(self):
    self.x += 5
    print("C:", self.x)
  def show(self):
    self.x += 4
    print("C show:", self.x)
    super().show()
class D(B, C):
  def __init__(self):
    super().__init__()
    self.x -= 1
    print("D:", self.x)
obj = D()
obj.show()
print("Final:", obj.x)
```

```python
class Person:
    def __init__(self, name):
        self.name = name


class Employee(Person):
    def __init__(self, name, emp_id, basic_salary):
        super().__init__(name)
        self.emp_id = emp_id
        self.basic_salary = basic_salary


class Manager(Employee):
    def __init__(self, name, emp_id, basic_salary, team_members):
        super().__init__(name, emp_id, basic_salary)
        self.team_members = team_members

    def calculate_salary(self):
        bonus = self.basic_salary * 0.15
        final_salary = self.basic_salary + bonus

        print("Name:", self.name)
        print("Employee ID:", self.emp_id)
        print("Basic Salary:", self.basic_salary)
        print("Team Members:", self.team_members)
        print("Management Bonus:", bonus)
        print("Final Salary:", final_salary)


m1 = Manager("Ali", 101, 80000, 10)
m1.calculate_salary()
```

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price


class ElectronicProduct(Product):
    def __init__(self, name, price, warranty):
        super().__init__(name, price)
        self.warranty = warranty


class Smartphone(ElectronicProduct):
    def __init__(self, name, price, warranty, os):
        super().__init__(name, price, warranty)
        self.os = os

    def calculate_discount(self):
        if self.price >= 100000:
            discount = self.price * 0.10
        else:
            discount = 0

        final_price = self.price - discount

        print("Product Name:", self.name)
        print("Price:", self.price)
        print("Warranty:", self.warranty, "years")
        print("Operating System:", self.os)
        print("Discount:", discount)
        print("Final Price:", final_price)


s1 = Smartphone("Samsung S25", 150000, 2, "Android")
s1.calculate_discount()
```

```python
class Course:
    def __init__(self, course_name, credit_hours):
        self.course_name = course_name
        self.credit_hours = credit_hours


class UndergraduateCourse(Course):
    def __init__(self, course_name, credit_hours, semester):
        super().__init__(course_name, credit_hours)
        self.semester = semester


class LabCourse(UndergraduateCourse):
    def __init__(self, course_name, credit_hours, semester, lab_hours):
        super().__init__(course_name, credit_hours, semester)
        self.lab_hours = lab_hours

    def display(self):
        total_hours = self.credit_hours + self.lab_hours

        print("Course Name:", self.course_name)
        print("Credit Hours:", self.credit_hours)
        print("Semester:", self.semester)
        print("Lab Hours:", self.lab_hours)
        print("Total Weekly Hours:", total_hours)

        if total_hours >= 5:
            print("Heavy Workload")
        else:
            print("Normal Workload")


c1 = LabCourse("Python Programming", 3, 2, 2)
c1.display()
```

```python
class Vehicle:
    def __init__(self, vehicle_name, daily_rate):
        self.vehicle_name = vehicle_name
        self.daily_rate = daily_rate


class Customer:
    def __init__(self, customer_name, customer_id):
        self.customer_name = customer_name
        self.customer_id = customer_id


class Rental:
    def __init__(self, vehicle, customer):
        self.vehicle = vehicle
        self.customer = customer

    def calculate_cost(self, days):
        original_cost = self.vehicle.daily_rate * days

        if original_cost >= 50000:
            discount = original_cost * 0.10
        else:
            discount = 0

        final_cost = original_cost - discount

        print("Customer Name:", self.customer.customer_name)
        print("Customer ID:", self.customer.customer_id)
        print("Vehicle Name:", self.vehicle.vehicle_name)
        print("Daily Rate:", self.vehicle.daily_rate)
        print("Number of Days:", days)
        print("Original Cost:", original_cost)
        print("Discount:", discount)
        print("Final Cost:", final_cost)


v1 = Vehicle("Toyota Corolla", 6000)
c1 = Customer("Ahmed", 201)

r1 = Rental(v1, c1)
r1.calculate_cost(10)
```

```python
class BankAccount:
    def __init__(self, account_no, balance):
        self.account_no = account_no
        self.balance = balance


class SavingsAccount(BankAccount):
    def __init__(self, account_no, balance, interest_rate):
        super().__init__(account_no, balance)
        self.interest_rate = interest_rate


class PremiumSavingsAccount(SavingsAccount):
    def __init__(self, account_no, balance, interest_rate, premium):
        super().__init__(account_no, balance, interest_rate)
        self.premium = premium

    def calculate_interest(self):
        if self.premium:
            effective_rate = self.interest_rate + 2
        else:
            effective_rate = self.interest_rate

        interest = self.balance * effective_rate / 100
        final_balance = self.balance + interest

        return effective_rate, interest, final_balance


class Customer:
    def __init__(self, name, account):
        self.name = name
        self.account = account

    def display(self):
        rate, interest, final_balance = (
            self.account.calculate_interest()
        )

        print("Customer Name:", self.name)
        print("Account Number:", self.account.account_no)
        print("Balance:", self.account.balance)
        print("Effective Interest Rate:", rate, "%")
        print("Interest Earned:", interest)
        print("Final Balance:", final_balance)


a1 = PremiumSavingsAccount(12345, 100000, 5, True)
c1 = Customer("Sara", a1)
c1.display()
```
